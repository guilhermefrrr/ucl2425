```python
import pandas as pd

pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', None)


# URL da página com a tabela de jogos
url = "https://fbref.com/en/comps/8/schedule/Champions-League-Scores-and-Fixtures"

# Lê a tabela HTML da página usando pandas com extract_links=True
dfs = pd.read_html(url, extract_links='body')

# Get the first dataframe from the list
df = dfs[0]

# Convert the 'Match Report' column to string type
df['Match Report'] = df['Match Report'].astype(str)

# Filter the dataframe and create copy
df = df[df['Match Report'].str.contains('Match Report', na=False)].copy()

# Extract links and assign to a new column
df.loc[:, 'URL'] = df['Match Report'].str.extract(r'(en/matches/\S+)')[0].str.rstrip(")'")

df = df.dropna(subset=['URL'])
df['URL'] = [f"https://fbref.com/{link}" for link in df['URL']]

df.reset_index(drop=True, inplace=True)

df
```

```python
dfs1 = pd.read_html(url)
df1 = dfs1[0]

df1 = df1[df1['Match Report'].str.contains('Match Report', na=False)]

df1 = df1.drop(columns=['Match Report', 'Notes'])

df1 = df1[df1['Wk'] != 'Wk']

df1.reset_index(drop=True, inplace=True)

df1
```
